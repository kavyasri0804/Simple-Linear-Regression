```python
X=df['YearsExperience']
y=df['Salary']
```

```python
X=X.values.reshape(-1,1)
X
```

```python
X.shape
```

```python
#split the data train and test
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.33,random_state=10)
```

```python
X_train
```

```python
y_train
```

```python
len(X_train),len(y_train)
```

```python
#Plot the training data
plt.figure(figsize=(5,3))
plt.title('SLR for training data')
plt.xlabel('X_train')
plt.ylabel('y_train')
plt.scatter(X_train,y_train,color='r',marker='*')
plt.show()
```

```python
#Giving data to Algorithm
from sklearn.linear_model import LinearRegression
reg=LinearRegression()
```

```python
reg.fit(X_train,y_train)
```

```python
#Finding m and c value
#m value
reg.coef_
```

```python
#c value
reg.intercept_
```

```python
#data->y=mx+c
#y=9514.52350354*x+25630.510121837833
```

```python
y_train_predictions=reg.predict(X_train)
```

```python
y_train_predictions
```

```python
y_train.shape
```

```python
X_train1=X_train.flatten()
X_train1
```

```python
train_data_comparision=pd.DataFrame({'X_train':X_train1,'y_train_actual':y_train,'y_train_predictions':y_train_predictions})
```

```python
train_data_comparision
```

```python
plt.figure(figsize=(5,3))
plt.title('SLR for training data')
plt.xlabel('X_train')
plt.ylabel('y_train')
plt.scatter(X_train,y_train,color='r',marker='*')
plt.plot(X_train,y_train_predictions,color='blue',marker='*')
plt.show()
```

```python
train_data_comparision.index
```

```python
numerator=0
denominator=0
for i in train_data_comparision.index:
  numerator=numerator+(train_data_comparision['y_train_actual'][i]-train_data_comparision['y_train_predictions'][i])**2
  denominator=denominator+(train_data_comparision['y_train_actual'][i]-train_data_comparision['y_train_actual'].mean())**2
print(f"Training Data Accuracy:{1-numerator/denominator}")
```

```python
from sklearn.metrics import r2_score
acc=r2_score(y_train,y_train_predictions)
acc
```

```python
#training loss using MSE
s=0
for i in train_data_comparision.index:
  s=s+(train_data_comparision['y_train_actual'][i]-train_data_comparision['y_train_predictions'][i])**2
print(f"Training Loss using MSE:{s/len(train_data_comparision)-1}")
```

```python
#training loss using built_in_function
from sklearn.metrics import mean_squared_error
loss=mean_squared_error(y_train,y_train_predictions)
loss
```

```python
#training loss using RMSE
s=0
for i in train_data_comparision.index:
  s=s+(train_data_comparision['y_train_actual'][i]-train_data_comparision['y_train_predictions'][i])**2
print(f"Training Loss using RMSE:{np.sqrt(s/len(train_data_comparision)-1)}")
```

```python
#training loss using built_in_function
from sklearn.metrics import root_mean_squared_error
loss=root_mean_squared_error(y_train,y_train_predictions)
loss
```

```python
#Test Performance
y_test_predictions=reg.predict(X_test)
```

```python
y_test_predictions
```

```python
X_test1=X_test.flatten()
X_test1
```

```python
test_data_comparision=pd.DataFrame({'X_test':X_test1,'y_test_actual':y_test,'y_test_predictions':y_test_predictions})
test_data_comparision
```

```python
from sklearn.metrics import r2_score
test_acc=r2_score(y_test,y_test_predictions)
print(f"Testing Accuracy:{test_acc}")
```

```python
from sklearn.metrics import root_mean_squared_error
test_loss=root_mean_squared_error(y_test,y_test_predictions)
print(f"Test loss:{test_loss}")
```
